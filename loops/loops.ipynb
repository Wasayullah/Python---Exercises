#### Print numbers from 1 to 10 using a for loop.

```python
for i in range(1,11):
    print(i)
```

#### Print all even numbers from 2 to 20.

```python
for i in range(2,21,2):
    print(i)
```

#### Calculate the sum of numbers from 1 to 100.

```python
sum = 0
for i in range(1,101):
    sum +=i
    print(sum)
```
