#### Print numbers from 1 to 10 using a for loop.

```python
for i in range(1,11):
    print(i)
```

#### Print all even numbers from 2 to 20.

```python
for i in range(2,21,2):
    print(i)
```

#### Calculate the sum of numbers from 1 to 100.

```python
sum = 0
for i in range(1,101):
    sum = sum+i
print("Sum of the numbers from 1 to 100 is :",sum)
```

#### Print numbers from 10 down to 1.

```python
for i in range(10,0,-1):
    print(i)
```

#### Take a number from the user and calculate its factorial.

```python
num = eval(input("Enter a number: "))
fact = 1
for i in range(1,(num+1)):
   fact = fact*i
print(f"Factorial of {num}: is {fact}")
```

#### Take a string and count how many vowels (a, e, i, o, u) it contains.

```python
vowels = ["a","e","i","o","u"]
string = input("Enter a string to calculate the number of vowels in it: ").lower().strip()
count = 0
for vowel in vowels:
        if vowel in string:
                count +=1
print(f"Number of vowels in {string.capitalize()} is: {count}")
```

#### Given a list, find the largest number without using max().

```python
numbers = [12, 45, 7, 89, 23]
greatest = numbers[0]
for number in numbers:
    if number >greatest :
        print(number)
        greatest = number
print(f"{greatest} is the greatest")
```
