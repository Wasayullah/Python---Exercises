#### Take a number from the user and calculate its factorial.

```python
num = eval(input("Enter a number: "))
fact = 1
for i in range(1,(num+1)):
   fact = fact*i
print(f"Factorial of {num}: is {fact}")
```

#### Take a string and count how many vowels (a, e, i, o, u) it contains.

```python
vowels = ["a","e","i","o","u"]
string = input("Enter a string to calculate the number of vowels in it: ").lower().strip()
count = 0
for vowel in vowels:
        if vowel in string:
                count +=1
print(f"Number of vowels in {string.capitalize()} is: {count}")
```

#### Given a list, find the largest number without using max().

```python
numbers = [12, 45, 7, 89, 23]
greatest = numbers[0]
for number in numbers:
    if number >greatest :
        greatest = number
print(f"{greatest} is the greatest")
```

#### Take a number from the user and determine whether it is prime using a for loop.

```python
num = eval(input("Enter a number :"))
is_Prime = True
if num<=1:
    is_Prime = False
else:

    for i in range(2,int(num**0.5)+1):
        if (num % i ==0 ):
          is_Prime = False
          break
if is_Prime:
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")
```

#### Print the first 10 numbers of the Fibonacci series.

```python
a, b = 0, 1
for _ in range(10):
    print(a, end=" ")
    a, b = b, a + b
```
