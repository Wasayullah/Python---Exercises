### Question:3
##### Find the area of a rectangle: length = 12, width = 5.

```python
length = 12
width = 5
area_of_rectangle = length*width
print(f"Area of the rectangle (with dimensions of {length} X {width}) is : {area_of_rectangle} unit")
```

### Question:4
#####  Convert 37 degrees Celsius to Fahrenheit (F = C*9/5+32).

```python
celsius = 37 
fahrenheit = (celsius * 9/5 )+32
print(f"{celsius}'C is equal to : {fahrenheit}'F")
```
